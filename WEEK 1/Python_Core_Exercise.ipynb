```python


#task 1.1
from datetime import datetime
import random

raw_sensor_data = []

DEVICE_TYPES = ['Temperature_Sensor', 'Humidity_Sensor', 'Pressure_Sensor', 'Light_Sensor']
LOCATIONS = ['Lab1', 'Lab2', 'OfficeA', 'OfficeB', 'Warehouse']

READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),
    'Humidity_Sensor': (30.0, 80.0),
    'Pressure_Sensor': (980.0, 1030.0),
    'Light_Sensor': (100, 1000)
}

for i in range(100):
    device_id = f"DEV{i+1:03}"

    timestamp = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)
    min_val, max_val = READING_RANGES[device_type]
    value = round(random.uniform(min_val, max_val), 2)

    reading = (timestamp, device_id, device_type, location, value)
    raw_sensor_data.append(reading)
for reading in raw_sensor_data:
    print(reading)

#task1.2

temperature_readings = []

for reading in raw_sensor_data:
    if reading[2] == 'Temperature_Sensor':
        timestamp = reading[0]
        device_id = reading[1]
        location = reading[3]
        temperature = reading[4]

        temp_dict = {
            'timestamp': timestamp,
            'device_id': device_id,
            'location': location,
            'temperature_celsius': temperature
        }

        temperature_readings.append(temp_dict)
for reading in temperature_readings:
  print(reading)

#task1.3

unique_locations = {reading[3] for reading in raw_sensor_data}
print(unique_locations)

#task1.4

device_stats = {}

for reading in raw_sensor_data:
    device_type = reading[2]
    value = reading[4]
    if device_type not in device_stats:
        device_stats[device_type] = (value, value)
    else:
        current_min, current_max = device_stats[device_type]
        new_min = min(current_min, value)
        new_max = max(current_max, value)
        device_stats[device_type] = (new_min, new_max)

print(device_stats)

#task1.5.1

sliced_readings_1 = raw_sensor_data[9:20]
sliced_readings_2 = raw_sensor_data[::5]
print(sliced_readings_2)

#task 1.5.2

sample_string = "Python Programming"
substring_1 = sample_string[7:19]
substring_2 = sample_string[5::-1]
print(substring_1)
print(substring_2)

#task 1.5.3

fifth_reading_device_type = raw_sensor_data[4][2]
device_type_suffix = fifth_reading_device_type[-6:]
print(fifth_reading_device_type)
print(device_type_suffix)

#task 2.1

def get_readings_by_location(readings_list, target_location):
  result = []

  for reading in readings_list:
    if reading[3] == target_location:
      result.append(reading)
  return(result)
filtered_readings = get_readings_by_location(raw_sensor_data, 'Lab1')
print(filtered_readings)

#task 2.2

def calculate_average_temperature(temp_data_list):
    total = 0
    count = 0

    for reading in temp_data_list:
        total += reading['temperature_celsius']
        count += 1

    if count == 0:
        return 0.0

    return total / count
avg_temp = calculate_average_temperature(temperature_readings)
print(avg_temp)

#task 2.3

def generate_report_line(reading_dict):
  return f"[{reading_dict['timestamp']}] Device {reading_dict['device_id']} at {reading_dict['location']}: {reading_dict['temperature_celsius']}°C"

for reading in temperature_readings:
    print(generate_report_line(reading))

#task 3.1

import os
reports_dir = 'sensor_reports'

if not os.path.exists(reports_dir):
    os.makedirs(reports_dir)
    print(f"\nDirectory '{reports_dir}' created.")
else:
    print(f"\nDirectory '{reports_dir}' already exists.")

#task 3.2

from datetime import datetime

if len(raw_sensor_data) > 1:
    first_timestamp_str = raw_sensor_data[0][0]
    last_timestamp_str = raw_sensor_data[-1][0]

    date_format = '%Y-%m-%d %H:%M:%S'

    first_dt = datetime.strptime(first_timestamp_str, date_format)
    last_dt = datetime.strptime(last_timestamp_str, date_format)

    duration = last_dt - first_dt

    days = duration.days

    hours = duration.seconds // 3600
    minutes = (duration.seconds % 3600) // 60

    print(f"Duration: {days} days, {hours} hours, {minutes} minutes")

#task 3.3

import random

for i in range(5):
    num = random.randint(100, 200)
    print(num)

#task 4.1

import os

report_filename = os.path.join(reports_dir, 'temperature_report.txt')

try:
    with open(report_filename, 'w') as f:
        for reading in temperature_readings:
            f.write(generate_report_line(reading) + '\n')

    print(f"\nTemperature report written to '{report_filename}'.")

except IOError as e:
    print(f"\nError writing file '{report_filename}': {e}")

#task 4.2

import os

report_filename = os.path.join('sensor_reports', 'temperature_report.txt')

try:
    with open(report_filename, 'r') as f:
        read_report_lines = f.readlines()

    print("First 3 lines from the report:")
    for line in read_report_lines[:3]:
        print(line.strip())

except IOError as e:
    print(f"Error reading file '{report_filename}': {e}")
```
