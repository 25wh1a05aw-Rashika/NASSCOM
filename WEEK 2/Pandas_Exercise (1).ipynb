```python
import pandas as pd
import numpy as np

# Task 1: Basic Pandas Functions and Converting Arrays to DataFrames

# 1.1 Create a Pandas Series
s1_1 = pd.Series([10, 20, 30, 40, 50])

# 1.2 Create a 3x4 NumPy array with random integers between 1 and 100
np_array1_2 = np.random.randint(1, 101, size=(3, 4))

# 1.3 Convert NumPy array into DataFrame
df1_3 = pd.DataFrame(np_array1_2, columns=['A', 'B', 'C', 'D'])

print(s1_1)
print(np_array1_2)
print(df1_3)
```

```python
# Task 2: Synthetic Data Generation for Practice

# Create the Date column
dates = pd.date_range(start='2023-01-01', end='2023-04-09', periods=100)

# Create the DataFrame
df_sales = pd.DataFrame({
    'Date': dates,
    'Region': np.random.choice(['East', 'West', 'North', 'South'], 100),
    'Product': np.random.choice(['Laptop', 'Mouse', 'Keyboard', 'Monitor'], 100),
    'Sales': np.random.randint(100, 1001, 100),
    'Quantity': np.random.randint(1, 11, 100)
})

# Introduce 5 random NaN values in Sales
sales_nan_indices = np.random.choice(df_sales.index, 5, replace=False)
df_sales.loc[sales_nan_indices, 'Sales'] = np.nan

# Introduce 3 random NaN values in Quantity
quantity_nan_indices = np.random.choice(df_sales.index, 3, replace=False)
df_sales.loc[quantity_nan_indices, 'Quantity'] = np.nan

print(df_sales)
```

```python
# Task 3: Indexing and Slicing in DataFrames

# 3.1 Select Region and Sales columns
df3_1 = df_sales[['Region', 'Sales']]

# 3.2 Select rows where Product is Laptop
df3_2 = df_sales[df_sales['Product'] == 'Laptop']

# 3.3 Select rows where Region is East AND Sales > 500
df3_3 = df_sales[
    (df_sales['Region'] == 'East') &
    (df_sales['Sales'] > 500)
]

# 3.4 Select Sales value for the first entry where Product is Monitor
monitor_index = df_sales.index[df_sales['Product'] == 'Monitor'][0]

val3_4 = df_sales.loc[monitor_index, 'Sales']

# Using iloc as well
position = df_sales.index.get_loc(monitor_index)
val3_4 = df_sales.iloc[position]['Sales']

print(df3_1)
print(df3_2)
print(df3_3)
print("First Monitor Sales:", val3_4)
```

```python
# Task 4: Data Cleaning (Handling Missing Values)

# 4.1 Check missing values in each column
missing_counts4_1 = df_sales.isna().sum()

# 4.2 Fill missing Sales values with mean Sales
df4_2_filled = df_sales.copy()

df4_2_filled['Sales'] = df4_2_filled['Sales'].fillna(
    df4_2_filled['Sales'].mean()
)

# 4.3 Drop rows having any missing values
df4_3_dropped = df_sales.dropna()

print("Missing values:")
print(missing_counts4_1)

print("\nDataFrame after filling Sales:")
print(df4_2_filled)

print("\nDataFrame after dropping missing rows:")
print(df4_3_dropped)
```

```python
# Task 5: Data Manipulation (Columns, Sorting, Grouping)

# 5.1 Create Total_Revenue column
df_sales['Total_Revenue'] = df_sales['Sales'] * df_sales['Quantity']

# 5.2 Sort by Date ascending, then Sales descending
df5_2_sorted = df_sales.sort_values(
    by=['Date', 'Sales'],
    ascending=[True, False]
)

# 5.3 Group by Region and calculate sum of Sales and Quantity
df5_3_grouped_region = df_sales.groupby('Region')[['Sales', 'Quantity']].sum()

# 5.4 Group by Product and calculate average Sales
df5_4_grouped_product = df_sales.groupby('Product')['Sales'].mean()

print("Sorted DataFrame:")
print(df5_2_sorted)

print("\nGrouped by Region:")
print(df5_3_grouped_region)

print("\nAverage Sales by Product:")
print(df5_4_grouped_product)
```
